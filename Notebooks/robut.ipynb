```python
from datasets import DatasetDict

# Assuming `datasets` is your loaded dataset
filtered_datasets = []

# Loop through each unique perturbation type
for p_type in set(datasets["perturbation_type"]):
    # Filter the dataset
    filtered = datasets.filter(lambda example: example["perturbation_type"] == p_type)
    # Select the first 20 examples
    filtered_datasets.append(filtered.select(range(min(20, len(filtered)))))

# Concatenate all filtered subsets into one dataset
from datasets import concatenate_datasets
final_dataset = concatenate_datasets(filtered_datasets)

# Optional: check result
print(final_dataset)
```
